```python
import yfinance as yf
import pandas as pd
```

```python
end_date = pd.Timestamp.today().normalize()

spx = yf.download(
    '^GSPC',
    start="1990-01-01",
    end=end_date,
    interval="1d"
)
qqq = yf.download(
    'QQQ',
    start="1990-01-01",
    end=end_date,
    interval="1d"
)
apple = yf.download(
    'AAPL',
    start="1990-01-01",
    end=end_date,
    interval="1d"
)
tesla = yf.download(
    'TSLA',
    start="1990-01-01",
    end=end_date,
    interval="1d"
)
btc = yf.download(
    "BTC-USD",
    start="1990-01-01",
    end=end_date,
    interval="1d"
)
```

```python
start = '2014-09-17'
end = '2026-05-29'
spx = spx.loc[start:end]
qqq = qqq.loc[start:end]
apple = apple.loc[start:end]
tesla = tesla.loc[start:end]
spx.to_parquet('SPX.parquet')
qqq.to_parquet('QQQ.parquet')
apple.to_parquet('Apple.parquet')
tesla.to_parquet('Tesla.parquet')
btc.to_parquet('BTC.parquet')
```
